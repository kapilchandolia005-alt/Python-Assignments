## Task 1: Calculate Area of a Circle

```python
def calculate_area(radius):
    pi = 3.14
    return pi * radius * radius

radius = float(input("Enter radius of circle: "))
print("Area of circle:", calculate_area(radius))
```

## Task 2: Check Palindrome

```python
def check_palindrome(text):
    text = text.lower().replace(" ", "")
    return text == text[::-1]

text = input("Enter a word or phrase: ")
if check_palindrome(text):
    print("It is a palindrome.")
else:
    print("It is not a palindrome.")
```

## Task 3: Count Vowels

```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for char in text.lower():
        if char in vowels:
            count += 1

    return count

text = input("Enter a string: ")
print("Number of vowels:", count_vowels(text))
```

## Task 4: Calculate Factorial

```python
def calculate_factorial(n):
    factorial = 1

    for i in range(1, n + 1):
        factorial *= i

    return factorial

n = int(input("Enter a non-negative integer: "))

if n >= 0:
    print("Factorial:", calculate_factorial(n))
else:
    print("Please enter a non-negative integer.")
```

## Task 5: Check Prime Number

```python
def check_prime(n):
    if n <= 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

n = int(input("Enter an integer: "))

if check_prime(n):
    print(n, "is a prime number.")
else:
    print(n, "is not a prime number.")
```

## Task 6: Lambda Function to Check Palindrome

```python
check_palindrome_lambda = lambda text: text.lower().replace(" ", "") == text.lower().replace(" ", "")[::-1]

text = input("Enter a string: ")

if check_palindrome_lambda(text):
    print("It is a palindrome.")
else:
    print("It is not a palindrome.")
```

## Task 7: Lambda Function for Rectangle Area

```python
rectangle_area = lambda length, width: length * width

length = float(input("Enter length: "))
width = float(input("Enter width: "))

print("Area of rectangle:", rectangle_area(length, width))
```

## Task 8: Lambda Function for Celsius to Fahrenheit

```python
celsius_to_fahrenheit = lambda c: (c * 9/5) + 32

celsius = float(input("Enter temperature in Celsius: "))

print("Temperature in Fahrenheit:", celsius_to_fahrenheit(celsius))
```

## Task 9: Reverse a String Using Slicing

```python
text = input("Enter a string: ")

reversed_text = text[::-1]

print("Reversed string:", reversed_text)
```

## Task 10: Count Occurrences of a Specific Word

```python
text = input("Enter a string: ")
word = input("Enter the word to count: ")

count = text.lower().split().count(word.lower())

print("Number of occurrences:", count)
```

## Task 11: Remove Whitespace Using Regular Expressions

```python
import re

text = input("Enter a string: ")

result = re.sub(r"\s+", "", text)

print("String without whitespace:", result)
```

## Task 12: Validate Email Address Using Regular Expressions

```python
import re

email = input("Enter your email address: ")

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

if re.match(pattern, email):
    print("Valid email address.")
else:
    print("Invalid email address.")
```
