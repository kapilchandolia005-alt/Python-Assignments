### 5. List vs Tuple

- List is mutable, Tuple is immutable
- List uses [ ], Tuple uses ( )
- List has more built-in methods (append, insert, remove...), Tuple has very few (count, index)
- Tuple is faster and takes less memory than list
- Tuple can be used as dictionary key (if elements are immutable), list can't
- List used when data changes, Tuple used for fixed data


### 6. Stack using List

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if self.is_empty():
            print("Stack empty")
            return
        return self.stack.pop()

    def is_empty(self):
        return len(self.stack) == 0

    def display(self):
        print(self.stack)


s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.display()
s.pop()
s.display()
print(s.is_empty())
```

### 8. Student Records using Dictionary

```python
students = {}

def add_student(roll, name, age, grade):
    students[roll] = {"name": name, "age": age, "grade": grade}

def update_student(roll, age=None, grade=None):
    if roll in students:
        if age:
            students[roll]["age"] = age
        if grade:
            students[roll]["grade"] = grade

def remove_student(roll):
    if roll in students:
        del students[roll]

def print_students():
    for roll, info in students.items():
        print(roll, info)


add_student(101, "Aditi", 16, "A")
add_student(102, "Rohan", 17, "B")
print_students()

update_student(102, age=18, grade="A")
print_students()

remove_student(101)
print_students()
```

### 9. Queue using List

```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        if self.is_empty():
            print("Queue empty")
            return
        return self.queue.pop(0)

    def is_empty(self):
        return len(self.queue) == 0

    def display(self):
        print(self.queue)


q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()
q.dequeue()
q.display()
print(q.is_empty())
```
